```python
%sql
SELECT 'customers' AS table_name, COUNT(*) AS total_rows
FROM finance_management.silver.customers

UNION ALL

SELECT 'vendors', COUNT(*)
FROM finance_management.silver.vendors

UNION ALL

SELECT 'chart_of_accounts', COUNT(*)
FROM finance_management.silver.chart_of_accounts

UNION ALL

SELECT 'revenue_transactions', COUNT(*)
FROM finance_management.silver.revenue_transactions

UNION ALL

SELECT 'expense_transactions', COUNT(*)
FROM finance_management.silver.expense_transactions

UNION ALL

SELECT 'invoices', COUNT(*)
FROM finance_management.silver.invoices

UNION ALL

SELECT 'budgets', COUNT(*)
FROM finance_management.silver.budgets

UNION ALL

SELECT 'journal_entries', COUNT(*)
FROM finance_management.silver.journal_entries;
```
