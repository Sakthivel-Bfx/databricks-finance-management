```python
%sql
SELECT
    COUNT(*) AS total_expense_records,
    COUNT(DISTINCT e.vendor_id) AS expense_vendors,
    COUNT(DISTINCT v.vendor_id) AS matched_vendors,
    SUM(CASE WHEN v.vendor_id IS NULL THEN 1 ELSE 0 END)
        AS unmatched_vendor_records
FROM finance_management.gold.fact_expense e
LEFT JOIN finance_management.gold.dim_vendor v
    ON e.vendor_id = v.vendor_id;
```

```python
%sql
SELECT
    v.vendor_id,
    v.vendor_name,
    v.vendor_category,
    e.currency,
    COUNT(*) AS transaction_count,
    SUM(e.amount) AS total_expense_amount
FROM finance_management.gold.fact_expense e
INNER JOIN finance_management.gold.dim_vendor v
    ON e.vendor_id = v.vendor_id
GROUP BY
    v.vendor_id,
    v.vendor_name,
    v.vendor_category,
    e.currency
ORDER BY total_expense_amount DESC;
```
