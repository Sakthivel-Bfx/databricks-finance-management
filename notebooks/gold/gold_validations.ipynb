```python
%sql
SELECT 'dim_customer' AS table_name, COUNT(*) AS record_count
FROM finance_management.gold.dim_customer

UNION ALL
SELECT 'dim_vendor', COUNT(*)
FROM finance_management.gold.dim_vendor

UNION ALL
SELECT 'dim_account', COUNT(*)
FROM finance_management.gold.dim_account

UNION ALL
SELECT 'fact_revenue', COUNT(*)
FROM finance_management.gold.fact_revenue

UNION ALL
SELECT 'fact_expense', COUNT(*)
FROM finance_management.gold.fact_expense

UNION ALL
SELECT 'fact_invoices', COUNT(*)
FROM finance_management.gold.fact_invoices

UNION ALL
SELECT 'fact_budgets', COUNT(*)
FROM finance_management.gold.fact_budgets

UNION ALL
SELECT 'fact_journal_entries', COUNT(*)
FROM finance_management.gold.fact_journal_entries;
```

```python
%sql
SELECT
    transaction_month,
    currency,
    total_revenue,
    total_expenses,
    gross_profit,
    total_revenue - total_expenses AS expected_profit
FROM finance_management.gold.profit_loss_summary
WHERE ABS(
    gross_profit - (total_revenue - total_expenses)
) > 0.01;
```
